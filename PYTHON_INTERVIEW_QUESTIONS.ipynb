**PYTHON INTERVIEW BASED QUESTIONS**

1. Reverse a string
Question: Write a Python program to reverse a string without using a built-in reverse function.

```python
text = "akash"
reversed_text = text[::-1]

print(reversed_text)
```

2. Check whether a string is a palindrome
Question: Check whether a word reads the same forward and backward.

```python
text = "madam"

if text == text[::-1]:
    print("Palindrome")
else:
    print("Not a palindrome")
```

3. Find the factorial of a number
Question: Calculate the factorial of 5 using a loop.

```python
n = 5
factorial = 1

for i in range(1, n + 1):
    factorial *= i

print(factorial)
```

4. Check whether a number is prime
Question: Determine whether a number is divisible only by 1 and itself, for numbers greater than 1.

```python
n = 17
is_prime = n > 1

for i in range(2, int(n ** 0.5) + 1):
    if n % i == 0:
        is_prime = False
        break

print("Prime" if is_prime else "Not prime")
```

5. Find the Fibonacci sequence
Question: Print the first 10 Fibonacci numbers, starting with 0 and 1.

```python
a, b = 0, 1

for _ in range(10):
    print(a, end=" ")
    a, b = b, a + b
```

6. Find the largest number in a list
Question: Find the largest number without using max().



```python
numbers = [12, 45, 7, 89, 23]
largest = numbers[0]

for num in numbers:
    if num > largest:
        largest = num

print(largest)
```

7. Find the second-largest distinct number
Question: Find the second-largest unique value in a list, including when the largest value occurs more than once.

```python
numbers = [10, 20, 40, 40, 30]

unique_numbers = sorted(set(numbers))
print(unique_numbers[-2])
```

8. Count the frequency of each character
Question: Count how many times each character occurs in a string.

```python
text = "banana"
frequency = {}

for char in text:
    frequency[char] = frequency.get(char, 0) + 1

print(frequency)
```

9. Remove duplicates from a list
Question: Remove duplicate elements while preserving their original order.

```python
numbers = [1, 2, 2, 3, 4, 3, 5]

unique = list(dict.fromkeys(numbers))

print(unique)
```

10. Check whether two strings are anagrams
Question: Determine whether two strings contain the same letters in different orders.

```python
s1 = "listen"
s2 = "silent"

if sorted(s1.lower()) == sorted(s2.lower()):
    print("Anagram")
else:
    print("Not an anagram")
```

11. Find the sum of digits of a number
Question: Calculate the sum of all digits in an integer.

```python
n = 12345
digit_sum = 0

while n > 0:
    digit_sum += n % 10
    n //= 10

print(digit_sum)
```

12. Count vowels in a string
Question: Count the vowels in a sentence, ignoring uppercase versus lowercase.

```python
text = "SABBA SAHIN"
vowels = "aeiou"

count = 0

for char in text.lower():
    if char in vowels:
        count += 1

print(count)
```

13. Swap two variables
Question: Swap two values without using a third variable.

```python
a = 10
b = 20

a, b = b, a

print("a =", a)
print("b =", b)
```

14. Find common elements between two lists
Question: Find the elements that occur in both lists without repeating results

```python
list1 = [1, 2, 3, 4, 5]
list2 = [3, 4, 5, 6, 7]

common = list(set(list1) & set(list2))

print(sorted(common))
```

15. Find the first non-repeating character
Question: Find the first character that occurs exactly once in a string.

```python
text = "swiss"

for char in text:
    if text.count(char) == 1:
        print(char)
        break
else:
    print("No non-repeating character")
```

16. Remove duplicates from a list
Interview question: Remove duplicate elements while maintaining their original order.

```python
numbers = [1, 2, 2, 3, 4, 3, 5]

unique = []

for num in numbers:
    if num not in unique:
        unique.append(num)

print("List without duplicates:", unique)
```
