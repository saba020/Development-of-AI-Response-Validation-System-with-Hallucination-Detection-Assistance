2. IMPORTING LIBRARIES

```python
import sqlite3
import pandas as pd
```

3.CREATING TABLE OF STUDENT, COURSES AND INSERTING VALUES


```python


# Create an in-memory SQLite database
connection = sqlite3.connect(":memory:")

# Create a cursor
cursor = connection.cursor()

# Enable foreign key enforcement
cursor.execute("PRAGMA foreign_keys = ON")

# Create tables and insert sample data
cursor.executescript("""
CREATE TABLE Students (
    student_id INTEGER PRIMARY KEY,
    student_name TEXT NOT NULL,
    age INTEGER,
    city TEXT,
    marks REAL
);

CREATE TABLE Courses (
    course_id INTEGER PRIMARY KEY,
    course_name TEXT NOT NULL,
    fee REAL NOT NULL
);

INSERT INTO Students VALUES
(1, 'Amit', 20, 'Kolkata', 85),
(2, 'Priya', 21, 'Delhi', 92),
(3, 'Rahul', 19, 'Mumbai', 76),
(4, 'Sneha', 22, 'Kolkata', 88),
(5, 'Arjun', 20, 'Chennai', 67),
(6, 'Neha', 21, 'Delhi', 95);

INSERT INTO Courses VALUES
(101, 'Python', 5000),
(102, 'SQL', 4000),
(103, 'Data Science', 8000),
(104, 'Web Development', 6000);
""")
def run(query):
    """Run a SQL query and show the result as a table."""
    return pd.read_sql_query(query, connection)

#
run("SELECT * FROM Students")
```

4.INSERTING NEW STUDENT IN STUDENT TABLE

```python
run("""
SELECT * FROM Students
WHERE student_id = 1
""")

cursor.execute("""
INSERT INTO Students
VALUES (7, 'Kiran', 23, 'Pune', 81)
""")
connection.commit()

run("SELECT * FROM Students WHERE student_id = 7;")
```

7.Find students scoring more than 80.

9. Find students aged between 20 and 22



```python
run("SELECT * FROM Students WHERE age BETWEEN 20 AND 22;")
```

	10.Count all students

```python
run("SELECT COUNT(*) AS total_students FROM Students;")
```

11. Count students in each city

```python
run("SELECT city, COUNT(*) AS total FROM Students GROUP BY city;")
```

12. LEFT JOIN — display all students and their courses.

```python
run("SELECT s.student_name, c.course_name FROM Students s LEFT JOIN Courses c ON s.student_id = c.course_id;")
```
