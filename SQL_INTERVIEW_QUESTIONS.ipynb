IMPORTING LIBRARIES

```python
import sqlite3
import pandas as pd
```

Connecting database

```python
import sqlite3

connection = sqlite3.connect("interview.db")
cursor = connection.cursor()
print("Database connected Successfully..")
```

1. Creating students table.

```python
cursor.execute("""
CREATE TABLE Students (
    student_id INTEGER PRIMARY KEY,
    name TEXT,
    department TEXT,
    marks INTEGER
)
""")

connection.commit()

print("Table created successfully!")
```

2. Inserting values in students table.


```python
cursor.execute("""
INSERT INTO Students (student_id, name, department, marks)
VALUES (1, 'Sabba', 'CSE', 85)
""")

cursor.execute("""
INSERT INTO Students (student_id, name, department, marks)
VALUES (2, 'Akash', 'CSE', 78)
""")

cursor.execute("""
INSERT INTO Students (student_id, name, department, marks)
VALUES (3, 'Surya', 'ECE', 90)
""")

cursor.execute("""
INSERT INTO Students (student_id, name, department, marks)
VALUES (4, 'Nafisha', 'BCA', 70)
""")

cursor.execute("""
INSERT INTO Students (student_id, name, department, marks)
VALUES (5, 'Ashish', 'CSE', 80)
""")

connection.commit()

print("Data inserted successfully!")
```

3. Showing data table.

```python
cursor.execute("""
SELECT * FROM Students;
""")

result = cursor.fetchall()

for row in result:
    print(row)
```

4. Display only names and marks.

```python
cursor.execute("""
SELECT name, marks
FROM Students;
""")

result = cursor.fetchall()

for row in result:
    print(row)
```

5. Find students who scored more than 70.

```python
cursor.execute("""
SELECT *
FROM Students
WHERE marks > 70;
""")

result = cursor.fetchall()

for row in result:
    print(row)
```

6. Find students who scored exatly 80.

```python
cursor.execute("""
SELECT *
FROM Students
WHERE marks = 80;
""")

result = cursor.fetchall()

for row in result:
    print(row)
```

7. Find CSE students.

```python
cursor.execute("""
SELECT *
FROM Students
WHERE department = 'CSE';
""")

result = cursor.fetchall()

for row in result:
    print(row)
```

8. Sort student by highest marks.

```python
cursor.execute("""
SELECT *
FROM Students
ORDER BY marks DESC;
""")

result = cursor.fetchall()

for row in result:
    print(row)
```

9. Sort student by lowest marks.

```python
cursor.execute("""
SELECT *
FROM Students
ORDER BY marks ASC;
""")

result = cursor.fetchall()

for row in result:
    print(row)
```

10. Find highest marks.

```python
cursor.execute("""
SELECT *
FROM Students
ORDER BY marks ASC;
""")

result = cursor.fetchall()

for row in result:
    print(row)
```

11. Find lowest marks.

```python
cursor.execute("""
SELECT MIN(marks) AS lowest_marks
FROM Students;
""")

result = cursor.fetchone()

print("Lowest marks:", result[0])
```
